## Remove irrelevant columns and process the data from the cities table

Remove the columns we don't need to keep the data clean.

```python
columns = ['amazonia_legal', 'id_uf','ddd','nome_regiao_metropolitana', 'id_regiao_metropolitana', 'id_mesorregiao','nome_microrregiao','id_microrregiao', 'nome_regiao_intermediaria', 'id_regiao_intermediaria', 'nome_regiao_imediata', 'id_regiao_imediata','nome_regiao_saude', 'id_regiao_saude', 'id_comarca', 'id_municipio_bcb', 'id_municipio_rf','id_municipio_tse', 'id_municipio_6', 'nome_mesorregiao', 'centroide']
data_cities = dt.drop_columns(data_cities,columns)
data_cities
```

Check if any column has missing values.

```python
data_missing = data_cities.isna().sum()
data_missing
```

Fill the missing `capital_uf` values with 0 (the city is not a capital).

```python
data_cities['capital_uf'] = data_cities['capital_uf'].fillna(0.0)
data_cities
```

Convert the `capital_uf` values from numbers to text (1 → "sim", 0 → "não").

```python
data_cities = prouni.change_data_capital(data_cities)
data_cities
```

## Prepare the join between the tables

Remove the `sigla_uf` column because it will come from the cities table after the join.

```python

ies_drop_colums = ['sigla_uf']
data_ies = dt.drop_columns(data_ies, ies_drop_colums);
data_ies
```

```python

maps = {"id_municipio": "id_municipio"}
df_ies = dt.join_data(data_ies, data_cities, maps, join_type='inner')
df_ies
```

Rename the columns of the IES table.

```python
columns_rename = { 
    "nome_x" : "instituicao",
    "id_municipio" : "id_municipio_ies",
    "nome_y": "nome_municipio_ies",
    "capital_uf": "capital_uf_ies",
    "sigla_uf": "uf_ies",
    "nome_uf": "nome_uf_ies",
    "nome_regiao": "nome_regiao_ies"
}

df_ies = df_ies.rename(columns=columns_rename)
df_ies
```

```python
drop_columns = ['sigla_uf','campus', 'nome_municipio_ies']
data_prouni = dt.drop_columns(data_prouni, drop_columns)
data_prouni
```

```python
maps = {"id_municipio": "id_municipio"}
df_prouni = dt.join_data(data_prouni, data_cities, maps, join_type='inner')
df_prouni
```

```python
columns_rename = { 
    "nome" : "cidade",
    "capital_uf" : "capital_estado",
    "nome_uf": "nome_estado",
    "sigla_uf": "uf"
}

df_prouni = df_prouni.rename(columns=columns_rename)
df_prouni
```

Join the Prouni and IES tables.

```python
print(f'Prouni list counter befofe join: {len(data_prouni)}')
```

```python
maps = {"id_ies": "id_ies"}
df_prouni = dt.join_data(df_prouni, df_ies, maps, join_type='inner')
df_prouni
```

```python
print(f'Prouni list counter after join: {len(data_prouni)}')
```

## Process the data and remove columns

Remove the columns we no longer need.

```python
drop_columns = ['id_municipio', 'id_ies', 'id_municipio_ies']
df_prouni = dt.drop_columns(df_prouni, drop_columns)
df_prouni
```

Convert the `data_nascimento` column to the date type.

```python
df_prouni['data_nascimento'] = pd.to_datetime(df_prouni['data_nascimento'], format='%Y-%m-%d')
df_prouni
```

```python
df_prouni.info()
```

Create a new column called `idade_candidato` with each candidate's age, calculated from `data_nascimento` and `ano` (the year the candidate applied to Prouni).

```python
df_prouni['idade_candidato'] = df_prouni['ano'] - df_prouni['data_nascimento'].dt.year
df_prouni
```
